# **HANDELING MISSING VALUES**

```python
df.dropna(subset = ['Customer_ID'], inplace = True)
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df['Age'].unique()
```

```python
import re

def extract_age(age):
  age_num = re.findall('[0-9]+', str(age))
  if len(age_num) > 0 :
    return age_num[0]
  else:
    return age

df['Age'] = df['Age'].apply(lambda x : extract_age(x))
```

```python
df['Age'].unique()
```

```python
df_age = df[df['Age'] != 'nan years']['Age']
age_median = int(df_age.dropna().astype('int64').median())
print(age_median)
df.replace('nan years', age_median, inplace = True)
df['Age'].fillna(age_median, inplace = True)
```

```python
df.isnull().sum()
```

```python
for col in ['City','Country','Gender']:
  df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
df['Purchase_Amount'].fillna(df['Purchase_Amount'].median(), inplace = True)
```

```python
df['Feedback_Score'].fillna(df['Feedback_Score'].mode()[0], inplace = True)
```

```python
df['Last_Purchase_Date'].ffill(inplace = True)
```

```python
df.isnull().sum()
```

# **FIXING INCONSISTENT FORMATING**

```python
df['Gender'].unique()
```

```python
df['Gender'] = df['Gender'].str.lower().str.strip()
```

```python
df['Gender'].replace({'f':'female','m':'male'}, inplace=True)
```

```python
df['Gender'].unique()
```

```python
df['City'].unique()
```

```python
df['City'] = df['City'].str.strip().str.lower()
```

```python
df['Country'].unique()
```

```python
df['Country'] = df['Country'].str.strip().str.lower()
```

```python
df['Country'].replace({'ind':'india'}, inplace=True)
```

# **HANDLING OUTLIERS**

```python
import numpy as np
from scipy import stats
```

```python
z_scores = np.abs(stats.zscore(df[['Age','Purchase_Amount']]))
threshold = 3
df_clean = df[~(z_scores > threshold).any(axis = 1)]
df_clean.shape
df_clean.describe()
```
